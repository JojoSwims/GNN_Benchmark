# Last-Value (Persistence) Baseline

Runs the `LastValueModel` on all five benchmark datasets.
For each test window the model predicts the last *observed* (non-NaN)
value in the input window for every horizon step.
No training is required; this serves as the canonical lower-bound baseline.
