# Last-Value (Persistence) Baseline

Runs the `LastValueModel` on all five benchmark datasets.
For each test window the model predicts the last *observed* (non-NaN)
value in the input window for every horizon step.
No training is required; this serves as the canonical lower-bound baseline.

```python
import sys
sys.path.insert(0, "..")

from gnn_benchmark.benchmark import BenchmarkRunner
from gnn_benchmark.models import LastValueModel

WORKSPACE = "../benchmark_workspace"
```

```python
DATASET = "noaa-buoy"
runner = BenchmarkRunner(workspace_dir=WORKSPACE, datasets=[DATASET])
result = runner.run(LastValueModel())
print(result.summary())
```
